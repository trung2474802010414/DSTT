Bài 1

```python
import numpy as np

# Vector A và B
A = np.array([3, -2, 1, 4])
B = np.array([-1, 5, 0, 2])

# a) Tổng và tích tuyến tính
sum_AB = A + B
linear_combination = 2 * A + 3 * B

print("a) A + B =", sum_AB)
print("2A + 3B =", linear_combination)

# b) Tích vô hướng và góc giữa 2 vector
dot_product = np.dot(A, B)
norm_A = np.linalg.norm(A)
norm_B = np.linalg.norm(B)
cos_theta = dot_product / (norm_A * norm_B)
angle_degree = np.degrees(np.arccos(cos_theta))

print("b) A · B =", dot_product)
print("Góc giữa A và B =", round(angle_degree, 3), "độ")

# c) Ma trận C nhân với C chuyển vị
C = np.array([[2, -1, 3],
              [1, 4, -2],
              [0, 2, 5]])

C_CT = C @ C.T
print("c) C * C^T =\n", C_CT)
```

Bài 2

```python
# Ma trận M và N
M = np.array([[2, -1, 3, 0],
              [1, 4, -2, 5],
              [0, 2, 6, -3],
              [-1, 0, 1, 4]])

N = np.array([[1, 3, -2, 0],
              [4, -1, 5, 2],
              [0, 2, 3, -1],
              [-3, 1, 0, 4]])

# a) Tích M × N
product_MN = M @ N
print("a) M × N =\n", product_MN)

# b) Định thức của M
det_M = np.linalg.det(M)
print("b) Định thức M =", det_M)

# c) Hạng của N
rank_N = np.linalg.matrix_rank(N)
print("c) Hạng của N =", rank_N)
```

Bài 3

```python
# Ma trận P
P = np.array([[3, -1, 0, 2],
              [1, 4, -2, 0],
              [0, 1, 5, -1],
              [2, 0, 1, 3]])

# Vector hằng số
b = np.array([6, 8, 3, 7])

# a) Ma trận nghịch đảo
try:
    P_inv = np.linalg.inv(P)
    print("a) Ma trận nghịch đảo P^-1 =\n", P_inv)
except np.linalg.LinAlgError:
    print("a) Ma trận P không khả nghịch")

# b) Giải hệ bằng nghịch đảo
x = P_inv @ b
print("b) Nghiệm x =", x)

# c) Kiểm tra lại nghiệm
check = P @ x
print("c) Kiểm tra P @ x =", check)
print("So sánh với b =", b)
```

Bài 4

```python
# Ma trận Q
Q = np.array([[5, -2, 1],
              [0, 3, -1],
              [2, 0, 4]])

# a) Giá trị riêng
eigenvalues, eigenvectors = np.linalg.eig(Q)
print("a) Giá trị riêng =", eigenvalues)

# b) Vector riêng
print("b) Vector riêng (theo cột):\n", eigenvectors)

# c) Chuẩn hóa vector riêng
norm_vectors = eigenvectors / np.linalg.norm(eigenvectors, axis=0)
print("c) Vector riêng đã chuẩn hóa:\n", norm_vectors)
```

Bài 5

```python
# Ma trận R và vector b
R = np.array([[4, -1, 2],
              [1, 3, -2],
              [0, 2, 5]])

b = np.array([12, 5, 9])

# a) Giải hệ phương trình
x = np.linalg.solve(R, b)
print("a) Nghiệm x =", x)

# b) Nghịch đảo R và so sánh nghiệm
try:
    R_inv = np.linalg.inv(R)
    x_check = R_inv @ b
    print("b) R^-1 =\n", R_inv)
    print("Nghiệm tính bằng R^-1 * b =", x_check)
    print("So sánh 2 nghiệm:", np.allclose(x, x_check))
except np.linalg.LinAlgError:
    print("R không khả nghịch")

# c) Chuẩn Frobenius
fro_norm = np.linalg.norm(R, 'fro')
print("c) Chuẩn Frobenius của R =", fro_norm)
```
