```python
import numpy as np

A = np.array([0, 0, -1, 2, 3, -1, 0, 1, -1, -1])
search_vector = np.array([0, 1, -1])

n = len(A)
k = len(search_vector)

# Tạo ma trận B chứa các đoạn con từ A
B = np.zeros((n - k + 1, k), dtype=int)
for i in range(n - k + 1):
    B[i] = A[i:i+k]

# Tính tích vô hướng giữa các hàng của B và search_vector
C = np.inner(B, search_vector)

# Kiểm tra và in các vị trí khớp
target = np.inner(search_vector, search_vector)
for i in range(len(C)):
    if C[i] == target:
        print(f"Vị trí {i}, đoạn khớp: {B[i]}")
```

```python
from PIL import Image, ImageDraw

# Mở ảnh và chuyển sang RGB nếu cần
input_image = Image.open('C:/traidat.PNG')
input_image = input_image.convert("RGB")
input_pixels = input_image.load()

# Tạo bộ lọc trung bình (Box blur)
box_kernel = [
    [1 / 9, 1 / 9, 1 / 9],
    [1 / 9, 1 / 9, 1 / 9],
    [1 / 9, 1 / 9, 1 / 9]
]
kernel = box_kernel
offset = len(kernel) // 2  # = 1

# Tạo ảnh đầu ra rỗng
output_image = Image.new("RGB", input_image.size)
draw = ImageDraw.Draw(output_image)

# Duyệt từng pixel (trừ viền)
for x in range(offset, input_image.width - offset):
    for y in range(offset, input_image.height - offset):
        acc = [0, 0, 0]  # R, G, B

        for a in range(len(kernel)):
            for b in range(len(kernel)):
                xn = x + a - offset
                yn = y + b - offset
                pixel = input_pixels[xn, yn]

                acc[0] += pixel[0] * kernel[a][b]  # R
                acc[1] += pixel[1] * kernel[a][b]  # G
                acc[2] += pixel[2] * kernel[a][b]  # B

        # Ghi giá trị đã làm mờ vào ảnh mới
        draw.point((x, y), (int(acc[0]), int(acc[1]), int(acc[2])))

# Lưu ảnh đã làm mờ
output_image.save('C:/traidat_lammo.PNG')
print("✅ Ảnh đã được làm mờ và lưu tại: C:/traidat_lammo.PNG")
```

```python
import numpy as np

# Ví dụ ma trận A (bạn thay bằng ma trận cụ thể của bạn nếu có)
A = np.array([
    [0.2, 0.2, 0.3, 0.3],
    [0.1, 0.3, 0.3, 0.3],
    [0.4, 0.2, 0.2, 0.2],
    [0.3, 0.3, 0.1, 0.3]
])

# Khởi tạo vector x ban đầu
x = np.array([1.0, 1.0, 1.0, 1.0])

# Thực hiện phép nhân A @ x một lần (tích ma trận)
result = np.dot(A, x)
print("Kết quả lần đầu A · x:\n", result)

print("\nLặp lại phép nhân A · x trong 10 lần:")
# Thực hiện lặp nhiều lần
x = np.array([1.0, 1.0, 1.0, 1.0])  # reset lại x ban đầu
for i in range(10):
    x = np.dot(A, x)
    print(f"Lần {i+1}: {x}")
```

```python
from PIL import Image, ImageDraw

# Mở ảnh
img = Image.open("C:/traidat.PNG").convert("RGB")
pixels = img.load()

# Bộ lọc Gaussian 5x5
gauss_kernel = [
    [1,  4,  6,  4, 1],
    [4, 16, 24, 16, 4],
    [6, 24, 36, 24, 6],
    [4, 16, 24, 16, 4],
    [1,  4,  6,  4, 1]
]
kernel = [[x / 256 for x in row] for row in gauss_kernel]
offset = len(kernel) // 2

# Ảnh kết quả
blurred = Image.new("RGB", img.size)
draw = ImageDraw.Draw(blurred)

for x in range(offset, img.width - offset):
    for y in range(offset, img.height - offset):
        acc = [0, 0, 0]
        for a in range(len(kernel)):
            for b in range(len(kernel)):
                xn = x + a - offset
                yn = y + b - offset
                pixel = pixels[xn, yn]
                for c in range(3):  # RGB
                    acc[c] += pixel[c] * kernel[a][b]
        draw.point((x, y), tuple(int(v) for v in acc))

blurred.save("C:/traidat_gauss_blur.png")
print("✅ Đã lưu ảnh làm mờ Gaussian tại: C:/traidat_gauss_blur.png")
```

```python
#bài 2
from PIL import Image, ImageDraw

# Mở ảnh gốc và chuyển sang grayscale
img = Image.open('C:/traidat.PNG').convert("L")  # Ảnh đen trắng
pixels = img.load()

# Bộ lọc rút xương (kernel a)
kernel_a = [
    [0, -1,  0],
    [-1, 4, -1],
    [0, -1,  0]
]

# Bộ lọc rút xương (kernel b)
kernel_b = [
    [-1, -1, -1],
    [-1,  8, -1],
    [-1, -1, -1]
]

def apply_kernel(image, kernel):
    width, height = image.size
    output = Image.new("L", image.size)
    draw = ImageDraw.Draw(output)
    offset = len(kernel) // 2

    for x in range(offset, width - offset):
        for y in range(offset, height - offset):
            acc = 0
            for a in range(len(kernel)):
                for b in range(len(kernel)):
                    xn = x + a - offset
                    yn = y + b - offset
                    acc += pixels[xn, yn] * kernel[a][b]
            acc = max(0, min(255, int(acc)))  # Giới hạn giá trị 0–255
            draw.point((x, y), acc)
    return output

# Áp dụng lần lượt 2 kernel
edge_a = apply_kernel(img, kernel_a)
edge_a.save("C:/traidat_edge_a.png")

edge_b = apply_kernel(img, kernel_b)
edge_b.save("C:/traidat_edge_b.png")

print("✅ Đã lưu ảnh làm nổi cạnh với kernel a và b tại: C:/traidat_edge_a.png, C:/traidat_edge_b.png")
```
